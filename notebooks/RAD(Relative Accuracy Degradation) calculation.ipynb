```python
def RAD(acc_raw_tx, acc_hybrid_tx, acc_raw_x, acc_hybrid_x):
  return (acc_raw_tx/acc_raw_x) - (acc_hybrid_tx/acc_hybrid_x)
```

```python
import pandas as pd

# --- Extracting Accuracy Values from comments ---

# VIT (Raw) Accuracies (from CQTFyO__cUyG) - these are the baseline 'raw' model
vit_accuracies = {
    'tiny': {'x': 89.22, 'tx': 85.49},
    'base': {'x': 95.52, 'tx': 94.87}
}

# Hybrid Accuracies (from 19UeAvbuQAx0)
hybrid_accuracies = {
    'tiny': {
        0.25: {'x': 60.12, 'tx': 53.22},
        0.5: {'x': 84.14, 'tx': 78.67},
        0.7: {'x': 87.62, 'tx': 83.73},
        0.9: {'x': 88.9, 'tx': 85.54}
    },
    'base': {
        0.25: {'x': 71.16, 'tx': 67.50},
        0.5: {'x': 93.62, 'tx': 92.17},
        0.7: {'x': 95.2, 'tx': 94.43},
        0.9: {'x': 95.36, 'tx': 94.85}
    }
}

# EViT Accuracies (from qeUMeQCwNBeZ)
evit_accuracies = {
    'tiny': {
        0.25: {'x': 44.06, 'tx': 37.6666667},
        0.5: {'x': 79.72, 'tx': 74.3933333},
        0.7: {'x': 86.92, 'tx': 83.0866667},
        0.9: {'x': 89.08, 'tx': 85.4133333}
    },
    'base': {
        0.25: {'x': 61.00, 'tx': 56.2666667},
        0.5: {'x': 92.48, 'tx': 91.0466667},
        0.7: {'x': 95.04, 'tx': 94.34},
        0.9: {'x': 95.54, 'tx': 94.8333333}
    }
}

# ToMe Accuracies (from BOl3q8C3Y6kx)
tome_accuracies = {
    'tiny': {
        0.25: {'x': None, 'tx': None}, # NA values
        0.5: {'x': 83.06, 'tx': 77.40},
        0.7: {'x': 86.86, 'tx': 82.73},
        0.9: {'x': 88.76, 'tx': 85.13}
    },
    'base': {
        0.25: {'x': None, 'tx': None}, # NA values
        0.5: {'x': 91.42, 'tx': 90.23},
        0.7: {'x': 94.64, 'tx': 93.83},
        0.9: {'x': 95.3, 'tx': 94.69}
    }
}

keep_rates = [0.25, 0.5, 0.7, 0.9]
configurations = ['tiny', 'base']

rad_results = []

for config in configurations:
    for kr in keep_rates:
        # Hybrid model accuracies (used as the 'hybrid' part of the comparison)
        acc_hybrid_tx = hybrid_accuracies[config][kr]['tx']
        acc_hybrid_x = hybrid_accuracies[config][kr]['x']

        # --- Calculate RAD for 'Raw vs Hybrid' ---
        # Here 'Raw' refers to the VIT model
        acc_raw_vit_tx = vit_accuracies[config]['tx']
        acc_raw_vit_x = vit_accuracies[config]['x']

        if acc_raw_vit_x != 0 and acc_hybrid_x != 0: # Ensure no division by zero
            rad_raw_hybrid = RAD(acc_raw_vit_tx, acc_hybrid_tx, acc_raw_vit_x, acc_hybrid_x)
            rad_results.append({
                'Comparison': 'Raw vs Hybrid',
                'Configuration': config,
                'Keep Rate': kr,
                'RAD': rad_raw_hybrid
            })
        else:
            rad_results.append({
                'Comparison': 'Raw vs Hybrid',
                'Configuration': config,
                'Keep Rate': kr,
                'RAD': float('nan')
            })

        # --- Calculate RAD for 'Tome vs Hybrid' ---
        # Here 'Tome' acts as the 'raw' model being compared against 'Hybrid'
        acc_tome_tx = tome_accuracies[config][kr]['tx']
        acc_tome_x = tome_accuracies[config][kr]['x']

        if acc_tome_tx is not None and acc_tome_x is not None and acc_tome_x != 0 and acc_hybrid_x != 0: # Check for NA and division by zero
            rad_tome_hybrid = RAD(acc_tome_tx, acc_hybrid_tx, acc_tome_x, acc_hybrid_x)
            rad_results.append({
                'Comparison': 'Tome vs Hybrid',
                'Configuration': config,
                'Keep Rate': kr,
                'RAD': rad_tome_hybrid
            })
        else:
            rad_results.append({
                'Comparison': 'Tome vs Hybrid',
                'Configuration': config,
                'Keep Rate': kr,
                'RAD': float('nan') # Represent missing data as NaN
            })

        # --- Calculate RAD for 'EViT vs Hybrid' ---
        # Here 'EViT' acts as the 'raw' model being compared against 'Hybrid'
        acc_evit_tx = evit_accuracies[config][kr]['tx']
        acc_evit_x = evit_accuracies[config][kr]['x']

        if acc_evit_x != 0 and acc_hybrid_x != 0: # Ensure no division by zero
            rad_evit_hybrid = RAD(acc_evit_tx, acc_hybrid_tx, acc_evit_x, acc_hybrid_x)
            rad_results.append({
                'Comparison': 'EViT vs Hybrid',
                'Configuration': config,
                'Keep Rate': kr,
                'RAD': rad_evit_hybrid
            })
        else:
            rad_results.append({
                'Comparison': 'EViT vs Hybrid',
                'Configuration': config,
                'Keep Rate': kr,
                'RAD': float('nan')
            })

# Create DataFrame
rad_df_full = pd.DataFrame(rad_results)
display(rad_df_full)
```
