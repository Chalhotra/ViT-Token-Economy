```python
model = timm.create_model('vit_tiny_patch16_224',pretrained=True)

device = "cuda" if torch.cuda.is_available() else "cpu"
```

```python
config = resolve_data_config({}, model=model)
transform = create_transform(**config)

def preprocess(example):
    # Convert RGB and apply timm transforms
    example["pixel_values"] = transform(example["image"].convert("RGB"))
    return example

ds = ds.map(preprocess, remove_columns=["image"])
ds.set_format(type="torch", columns=["pixel_values", "label"])
```

```python
original_weight = model.head.weight.data      # [1000, 768]
original_bias = model.head.bias.data

new_weight = torch.zeros((100, model.head.in_features))
new_bias = torch.zeros(100)

for new_idx, old_idx in new_to_old_map.items():
    new_weight[new_idx] = original_weight[old_idx]
    new_bias[new_idx] = original_bias[old_idx]

model.head = torch.nn.Linear(model.head.in_features, 100)
model.head.weight.data = new_weight
model.head.bias.data = new_bias

model = model.to(device)
model = model.eval()
```

```python
loader = DataLoader(
    ds,
    batch_size=64,
    shuffle=True,
    num_workers=4,
    pin_memory=True,
)
```

```python
correct = 0
total = 0
total_time = 0.0

# token stats
total_tokens = 0
num_blocks = len(model.blocks)
i = 0

with torch.no_grad():
    for batch in loader:
        images = batch["pixel_values"].to(device)
        labels = batch["label"].to(device)

        # sync before timing (important on GPU)
        if device == "cuda":
            torch.cuda.synchronize()

        start = time.perf_counter()
        outputs = model(images)
        if device == "cuda":
            torch.cuda.synchronize()
        end = time.perf_counter()

        # time
        total_time += (end - start)

        # accuracy
        preds = outputs.argmax(dim=1)
        correct += (preds == labels).sum().item()
        total += labels.size(0)
        print(f'{i+1}/79 batches completed')
        i+=1

avg_latency = (total_time / total) * 1000 if total > 0 else 0 # ms per image
throughput = total / total_time if total_time > 0 else 0     # images per sec
accuracy = (correct / total) * 100 if total > 0 else 0

print(f"Accuracy: {accuracy:.2f}%")
print(f"Throughput: {throughput:.2f} im/s")
print(f"Latency: {avg_latency:.2f} ms/image")
```
